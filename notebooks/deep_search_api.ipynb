```python
import requests
from datetime import datetime
from dateutil.relativedelta import relativedelta
from tqdm import tqdm
API_KEY = 'a6f13e22dcc6488f8bc64b5e8ea8ceed'
date = datetime(2018, 7, 1)
articles = []

for i in tqdm(range(1815)):
    start_date = date + relativedelta(weeks=i)
    end_date = start_date + relativedelta(weeks=1)

    url = f"https://api-v2.deepsearch.com/v1/articles/economy?page=1&page_size=100&keyword=(아파트 OR 매매)&date_from={start_date.strftime('%Y-%m-%d')}&date_to={end_date.strftime('%Y-%m-%d')}&api_key={API_KEY}"

    try:
        response = requests.get(url)
        if response.status_code == 200:
            data = response.json()
            #print(f"{start_date.strftime('%Y-%m-%d')} 데이터 요청 성공, 기사 개수: {len(data['data'])}")

            for item in data['data']:
                article = {
                    "기사 ID": item['id'],
                    "기사 제목": item['title'],
                    "출판사": item['publisher'],
                    "게시 일자": item['published_at'],
                    "URL": item['content_url'],
                    "본문": item['summary']
                }
                articles.append(article)
        else:
            print(f"요청 실패 : {response.status_code}")
            print(f"응답 내용 : {response.text}")

    except requests.exceptions.RequestException as e:
        print(f"요청 중 오류 발생 : {e}")
```
