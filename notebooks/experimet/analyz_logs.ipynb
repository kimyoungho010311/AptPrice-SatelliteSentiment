```python
import re
from collections import defaultdict
from datetime import datetime

# 로그 파일 경로
log_path = "../../data/log/image_download/image_download.log"

# 로그 문자열 읽기
with open(log_path, "r", encoding="utf-8") as f:
    log = f.read()

# 정규표현식: timestamp + endpoint + 429 감지
pattern = r"\[(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}),\d+\].+?POST.+?/([^/?]+).*?429"

requests_per_second = defaultdict(int)

# 초 단위 요청 수 카운트
for match in re.finditer(pattern, log):
    timestamp_str, endpoint = match.groups()
    timestamp = datetime.strptime(timestamp_str, "%Y-%m-%d %H:%M:%S")
    requests_per_second[timestamp] += 1

# 결과 출력: 429 발생 시점의 초당 요청 수
for t, count in sorted(requests_per_second.items()):
    print(f"{t} → 요청 수: {count}")
```
