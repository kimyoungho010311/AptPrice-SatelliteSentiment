```python
import pandas as pd
import os
import warnings
import datetime
warnings.filterwarnings("ignore", category=UserWarning, module="openpyxl")
```

```python
folder_path = "../data/raw/news"
```

```python
# 폴더 내 모든 .xlsx 파일 경로 리스트업
xlsx_files = [os.path.join(folder_path, file)
              for file in os.listdir(folder_path)
              if file.endswith('.xlsx')]
```

```python
# 파일들을 DataFrame으로 읽어서 리스트에 저장
df_list = [pd.read_excel(file) for file in xlsx_files]
```

```python
# 하나의 DataFrame으로 병합
merged_df = pd.concat(df_list, ignore_index=True)
```

```python
merged_df.columns
```

```python
df = merged_df[['일자','언론사','본문']]
```

```python
df['일자'] = pd.to_datetime(df['일자'], format='%Y%m%d')
```

```python
df.rename(columns={'일자':'date', '언론사':'press', '본문':'content'},inplace=True)
```
